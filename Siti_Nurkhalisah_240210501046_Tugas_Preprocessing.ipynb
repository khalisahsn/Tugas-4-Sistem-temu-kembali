```python
!pip install Sastrawi

import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory


stopword = StopWordRemoverFactory().get_stop_words()
stemmer = StemmerFactory().create_stemmer()


def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    tokens_before = text.split()

    tokens_stopword = [
        token for token in tokens_before
        if token not in stopword
    ]

    tokens_after = [
        stemmer.stem(token)
        for token in tokens_stopword
    ]

    text_processed = ' '.join(tokens_after)

    return text_processed, tokens_before, tokens_after


dokumen = [
    "Konsep transformasi digital dalam pendidikan tinggi berbasis teknologi informasi.",
    "Pemanfaatan teknologi digital dalam meningkatkan kualitas pendidikan di perguruan tinggi.",
    "Penerapan teknologi informasi dalam pengelolaan data akademik mahasiswa.",
    "Pengembangan jaringan komputer untuk mendukung akses informasi di lingkungan kampus.",
    "Pemanfaatan kecerdasan buatan dalam mendukung proses pembelajaran di perguruan tinggi."
]


for i, doc in enumerate(dokumen, start=1):

    hasil, tokens_before, tokens_after = preprocess_text(doc)

    jumlah_sebelum = len(tokens_before)
    jumlah_sesudah = len(tokens_after)

    if jumlah_sebelum > 0:
        persen_pengurangan = (
            (jumlah_sebelum - jumlah_sesudah)
            / jumlah_sebelum
        ) * 100
    else:
        persen_pengurangan = 0

    print(f"================ DOKUMEN {i} ================")

    print("\nSebelum preprocessing:")
    print(doc)

    print("\nSesudah preprocessing:")
    print(hasil)

    print("\nToken sebelum:")
    print(tokens_before)

    print("\nToken sesudah:")
    print(tokens_after)

    print("\nJumlah token sebelum:", jumlah_sebelum)
    print("Jumlah token sesudah:", jumlah_sesudah)
    print(
        "Persentase pengurangan:",
        f"{persen_pengurangan:.2f}%"
    )

    print("\n" + "=" * 55 + "\n")
```

**Analisis:**

Preprocessing teks berperan dalam membuat data menjadi lebih teratur sehingga lebih mudah diproses oleh sistem Information Retrieval (IR). Tahapannya meliputi mengubah seluruh huruf menjadi huruf kecil, membersihkan angka dan tanda baca, serta memisahkan teks menjadi beberapa token. Setelah itu, kata-kata yang umum dan kurang penting dihilangkan, kemudian kata berimbuhan diubah menjadi bentuk dasarnya. Proses ini membuat jumlah token lebih ringkas dan membantu sistem mengenali kata-kata yang memiliki bentuk berbeda tetapi berasal dari kata dasar yang sama, sehingga pencarian dokumen dapat dilakukan dengan lebih baik.
